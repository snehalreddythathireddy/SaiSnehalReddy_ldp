```python
from sentence_transformers import SentenceTransformer
from sentence_transformers import util
import time
```

```python
sentences = [
    "I love learning Artificial Intelligence.",
    "I enjoy studying Machine Learning.",
    "Cricket is my favourite sport.",
    "Deep Learning is a part of Machine Learning.",
    "I like playing football."
]
```

```python
models = [
    "sentence-transformers/all-MiniLM-L6-v2",
    "sentence-transformers/all-mpnet-base-v2",
    "sentence-transformers/multi-qa-MiniLM-L6-cos-v1",
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
]

for model_name in models:

    print("\nModel:", model_name)

    # Load model
    model = SentenceTransformer(model_name)

    # Generate embeddings
    start_time = time.time()

    embeddings = model.encode(sentences)

    end_time = time.time()

    # Display results
    print("Embedding shape:", embeddings.shape)
    print("Execution time:", round(end_time - start_time, 3), "seconds")
    print("First sentence embedding:")
    print(embeddings[0][:10])
```

```python
from sentence_transformers import SentenceTransformer, util

sentence1 = "I love learning Artificial Intelligence."
sentence2 = "I enjoy studying Machine Learning."

models = [
    "sentence-transformers/all-MiniLM-L6-v2",
    "sentence-transformers/all-mpnet-base-v2",
    "sentence-transformers/multi-qa-MiniLM-L6-cos-v1",
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
]

for model_name in models:

    model = SentenceTransformer(model_name)

    embeddings = model.encode([sentence1, sentence2])

    similarity = util.cos_sim(
        embeddings[0],
        embeddings[1]
    )

    print(model_name)
    print("Similarity:", round(similarity.item(), 4))
    print("-" * 50)
```
