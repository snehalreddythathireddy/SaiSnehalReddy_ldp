```python

from sentence_transformers import SentenceTransformer, util

# Load embedding model
model = SentenceTransformer("Qwen/Qwen3-Embedding-0.6B")

# Course dataset
courses = [
    "Introduction to Artificial Intelligence and intelligent systems.",
    "Python programming for beginners and software development.",
    "Machine Learning algorithms for prediction and classification.",
    "Web development using HTML, CSS, JavaScript and React.",
    "Database management using SQL and relational databases.",
    "Deep Learning with neural networks and computer vision.",
    "Cybersecurity fundamentals and network protection.",
    "Data Science with Python, statistics and data visualization."
]

# Generate course embeddings
course_embeddings = model.encode(
    courses,
    convert_to_tensor=True,
    normalize_embeddings=True
)

# User interest
user_query = "I want to learn how to build intelligent systems that make predictions."

# Generate query embedding
query_embedding = model.encode(
    user_query,
    convert_to_tensor=True,
    normalize_embeddings=True
)

# Calculate similarity
similarity_scores = util.cos_sim(
    query_embedding,
    course_embeddings
)[0]

# Retrieve top 3 courses
top_results = similarity_scores.argsort(descending=True)[:3]

# Display recommendations
print("USER INTEREST:", user_query)
print("\nRECOMMENDED COURSES:\n")

for index in top_results:
    index = index.item()

    print("Course:", courses[index])
    print("Similarity:", round(similarity_scores[index].item(), 4))
    print("-" * 50)

print("Embedding Shape:", course_embeddings.shape)
print("Embedding Dimensions:", course_embeddings.shape[1])
```
